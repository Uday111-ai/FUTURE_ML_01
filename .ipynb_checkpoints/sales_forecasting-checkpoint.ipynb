# Sales & Demand Forecasting Using Machine Learning

## Objective
The goal of this project is to predict future sales based on historical retail data and present insights in a business-friendly way to support decision-making.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
# Load dataset
df = pd.read_csv("../data/raw/Sample - Superstore.csv", encoding="latin1")
df.head()
```

```python
# Convert to datetime
df['Order Date'] = pd.to_datetime(df['Order Date'], errors='coerce')
```

```python
# Drop missing dates
df = df.dropna(subset=['Order Date'])
```

```python
# Monthly Aggregation

monthly_sales = df.groupby(pd.Grouper(key='Order Date', freq='ME'))['Sales'].sum().reset_index()
monthly_sales = monthly_sales.sort_values('Order Date')
```

```python
# Feature Engineering

for lag in [1,2,3,6,12]:
    monthly_sales[f'lag_{lag}'] = monthly_sales['Sales'].shift(lag)

monthly_sales['rolling_mean_3'] = monthly_sales['Sales'].shift(1).rolling(3).mean()
monthly_sales['rolling_mean_6'] = monthly_sales['Sales'].shift(1).rolling(6).mean()

monthly_sales['month'] = monthly_sales['Order Date'].dt.month
monthly_sales['quarter'] = monthly_sales['Order Date'].dt.quarter
monthly_sales['year'] = monthly_sales['Order Date'].dt.year

monthly_sales = monthly_sales.dropna()
```

## Time-Based Split

```python
split_date = '2017-01-01'

train = monthly_sales[monthly_sales['Order Date'] < split_date]
test  = monthly_sales[monthly_sales['Order Date'] >= split_date]

X_train = train.drop(['Sales','Order Date'], axis=1)
y_train = train['Sales']

X_test = test.drop(['Sales','Order Date'], axis=1)
y_test = test['Sales']
```

## Train Model

```python
model = RandomForestRegressor(
    n_estimators=600,
    max_depth=25,
    random_state=42,
    n_jobs=-1
)

model.fit(X_train, y_train)
predictions = model.predict(X_test)
```

```python
# Predictions

predictions = model.predict(X_test)
```

```python
# Evaluation

mae = mean_absolute_error(y_test, predictions)
rmse = np.sqrt(mean_squared_error(y_test, predictions))
r2 = r2_score(y_test, predictions)

print("MAE:", mae)
print("RMSE:", rmse)
print("R2:", r2)
```

```python
# Visualization

plt.figure(figsize=(12,6))
plt.plot(test['Order Date'], y_test, label='Actual Sales')
plt.plot(test['Order Date'], predictions, label='Predicted Sales')
plt.title("Sales Forecast vs Actual")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

```python
# Feature Importance

feature_importance = pd.Series(
    model.feature_importances_,
    index=X_train.columns
).sort_values(ascending=False)

print("\nTop Important Features:")
print(feature_importance.head(10))
```

```python
plt.figure(figsize=(12,6))

plt.plot(test['Order Date'], y_test, marker='o', label='Actual Sales')
plt.plot(test['Order Date'], predictions, marker='s', linestyle='--', label='Predicted Sales')

plt.title("Monthly Sales Forecast vs Actual")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
quarterly = monthly_sales.copy()
quarterly['Quarter'] = quarterly['Order Date'].dt.quarter

quarter_summary = quarterly.groupby('Quarter')['Sales'].mean().reset_index()

plt.figure(figsize=(8,5))
plt.bar(quarter_summary['Quarter'], quarter_summary['Sales'])

plt.title("Average Quarterly Sales")
plt.xlabel("Quarter")
plt.ylabel("Average Sales")
plt.show()
```

```python
# ==============================
# 1️⃣ Store Training Feature Order
# ==============================

feature_columns = X_train.columns


# ==============================
# 2️⃣ Recursive 6-Month Forecast
# ==============================

future_df = monthly_sales.copy()
future_predictions = []

for i in range(6):

    # Last available row
    last_row = future_df.iloc[-1]

    # Create next month date
    next_date = last_row['Order Date'] + pd.DateOffset(months=1)

    # Create new row dictionary
    new_row = {}

    new_row['Order Date'] = next_date

    # ---- Time Features ----
    new_row['year'] = next_date.year
    new_row['month'] = next_date.month
    new_row['quarter'] = (next_date.month - 1)//3 + 1

    # ---- Lag Features ----
    for lag in [1,2,3,6,12]:
        new_row[f'lag_{lag}'] = future_df['Sales'].iloc[-lag]

    # ---- Rolling Features ----
    new_row['rolling_mean_3'] = future_df['Sales'].iloc[-3:].mean()
    new_row['rolling_mean_6'] = future_df['Sales'].iloc[-6:].mean()

    # Convert to DataFrame
    new_row_df = pd.DataFrame([new_row])

    # Align feature order EXACTLY as training
    X_new = new_row_df[feature_columns]

    # Predict
    prediction = model.predict(X_new)[0]

    # Add predicted sales to row
    new_row_df['Sales'] = prediction

    # Append to future_df so next month can use this prediction
    future_df = pd.concat([future_df, new_row_df], ignore_index=True)

    future_predictions.append(new_row_df)


# Combine forecast results
future_forecast = pd.concat(future_predictions)

# ==============================
# 3️⃣ Show Forecast Table
# ==============================

print("Next 6 Months Forecast:")
print(future_forecast[['Order Date','Sales']])
```
